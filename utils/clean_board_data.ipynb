```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/original_board.csv")
```

```python
new_columns = str(df.columns[0]).split("|")
new_columns
```

```python
df.shape
```

```python
for col_indx, col_name in enumerate(new_columns):
    df[col_name] = str
    for x in range(df.shape[0]):
        df.loc[x, col_name] = str(df.loc[x, df.columns[0]]).split("|")[col_indx]
```

```python
df.replace("Street", "Property", inplace=True)
```

```python
df.drop(df.columns[0], inplace=True, axis=1)
```

```python
df
```

```python
df.to_csv("../data/board.csv", index=False)
```
